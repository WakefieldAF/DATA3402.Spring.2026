1. Write a "counter" class that can be incremented up to a specified maximum value, will print an error if an attempt is made to increment beyond that value, and allows reseting the counter. 

```python
class Counter:
    def __init__(self, max_val):
        self.max_val = max_val
        self.value = 0

    def increment_value(self, amount = 1):
        if self.value + amount > self.max_val:
            print("Error: Reached max value.", self.max_val)
        else:
            self.value += amount

    def reset_counter(self):
        self.value = 0

counter = Counter(10)

print("Counter is counting to 10: Starting counter..\n")
print("Counter value is ", counter.value)

while counter.value < counter.max_val:
    counter.increment_value()
    print("Counter value is ", counter.value)    

    if counter.value == counter.max_val:
        get_input = input("Press y to reset the counter or n to exit.\n")

        if get_input == "y":
            counter.reset_counter()

            print("Counter reset.\n")
            print("Counter value is ", counter.value)
        else:
            print("Exiting Counter.")
            break
```

Because right now your save() is writing something to the file… but it’s not writing “instructions to rebuild your drawing.” It’s writing “mystery object labels” that Python can’t use later.

Think of it like this:

Kid version

A shape object (Rectangle/Circle/Triangle) is like a toy.

When you “save,” you don’t want to save the toy itself (you can’t shove a real toy into a text file).

You want to save the recipe for building the toy again later.

A recipe looks like:

Rectangle(10, 6, 2, 2, name="Rectangle1", char="^")

But if your shape does not have a proper __repr__, Python writes something useless like:

<__main__.Rectangle object at 0x7f9a2c1c9d90>

That’s not a recipe. That’s like writing:

“A toy exists somewhere on my floor.”

When you later try to load it with eval(), Python reads that and says:

“I can’t build a Rectangle from ‘object at 0x…’.”

What’s happening in your code (simple example)
What you want to save

If your drawing has 2 shapes, you want the file to contain something like:

RasterDrawing([Rectangle(5,5,10,5), Circle(3,6,15)], None)

That is real Python code. eval() can run it and rebuild the drawing.

What you’re actually saving right now (without shape __repr__)

Your RasterDrawing.__repr__() does this:

repr(self.a)

But self.a is a list of shape objects. If shapes don’t have __repr__, that becomes:

[<__main__.Rectangle object at 0x...>, <__main__.Circle object at 0x...>]

So the file becomes:

RasterDrawing([<__main__.Rectangle object at 0x...>, <__main__.Circle object at 0x...>], None)

That is NOT valid Python code. So loading fails.

Make every shape able to write its own recipe.
