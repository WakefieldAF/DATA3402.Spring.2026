*Exercise 1:* Write a function that creates an n by n matrix (of list of lists) which will represent the state of a Tie Tac Toe game. Let 0, 1, and 2 represent empty, "X", and "O", respectively.


````python
def make_board(num):
    try:
        if type(num) != int:
            print("Error")
            return False
        if num <= 0:
            print("Error")
            return False
    except:
        print("Error")
        return False

    board = []
    row = 0

    while row < num:
        board.append([])
        column = 0
        while column < num:
            board[row].append(0)
            column += 1
        row += 1
    return board
````

*Exercise 2:* Write a function that takes 2 integers `n` and `m` as input and draws a `n` by `m` game board. For example the following is a 3x3 board:
```
   --- --- --- 
  |   |   |   | 
   --- --- ---  
  |   |   |   | 
   --- --- ---  
  |   |   |   | 
   --- --- --- 
   ```

````python
def show_board(n, m):
    try:
        if type(n) != int or type(m) != int:
            print("Error")
            return False
        if n <= 0 or m <= 0:
            print("Error")
            return False
    except:
        print("Error")
        return False

    vertical_cell = 1
    horizontal_cell = "  " + " ".join("-" * (vertical_cell + 2) for _ in range(m))
    print(horizontal_cell)
    
    row = 0

    while row < n:
        line = "  | " + " | ".join(" " * vertical_cell for _ in range(m)) + " |"
        print(line)
        print(horizontal_cell)
        row += 1
    return True

show_board(3,3)
````

*Exercise 3:* Modify exercise 2, so that it takes a matrix of the form from exercise 1 and draws a tic-tac-tie board with "X"s and "O"s.  

````python
def show_board(tac_grid):
    num = len(tac_grid)

    vertical_cell = 1
    horizontal_cell = "  " + " ".join("-" * (vertical_cell + 2) for _ in range(num))
    print(horizontal_cell)

    row = 0

    while row < num:
        shown_row = []
        column = 0
        while column < num:
            if tac_grid[row][column] == 1:
                shown_row.append("X")
            elif tac_grid[row][column] == 2:
                shown_row.append("O")
            else:
                shown_row.append(" ")
                
            column += 1
        lines = "  | " + " | ".join(shown_row) + " |"
        print(lines)
        print(horizontal_cell)
        row += 1
    return True

winner_is_2 = [[2, 2, 0],
	[2, 1, 0],
	[2, 1, 1]]

winner_is_1 = [[1, 2, 0],
	[2, 1, 0],
	[2, 1, 1]]

winner_is_also_1 = [[0, 1, 0],
	[2, 1, 0],
	[2, 1, 1]]

no_winner = [[1, 2, 0],
	[2, 1, 0],
	[2, 1, 2]]

also_no_winner = [[1, 2, 0],
	[2, 1, 0],
	[2, 1, 0]]

show_board(winner_is_1)
````

````python
def player_move(board, player, x, y):
    num = len(board)

    if x < 0:
        return False
    if x >= num:
        return False
    if y < 0:
        return False
    if y >= num:
        return False

    if board[x][y] != 0:
        return False

    board[x][y] = player
    return True        
````

*Exercise 9:* Use all of the previous exercises to implement a full tic-tac-toe game, where an appropriate board is drawn, 2 players are repeatedly asked for a location coordinates of where they wish to place a mark, and the game status is checked until a player wins or a draw occurs.

````python
def make_board(num):
    try:
        if type(num) != int:
            print("Error")
            return False
        if num <= 0:
            print("Error")
            return False
    except:
        print("Error")
        return False

    board = []
    row = 0

    while row < num:
        board.append([])
        column = 0
        while column < num:
            board[row].append(0)
            column += 1
        row += 1
    return board

def check_state(board):
    num = len(board)

    check = 0
    index = 0
    while check < num:
        if board[check][0] != 0 and all(board[check][row] == board[check][0] for row in range(num)):
            return board[check][0]

        if board[0][check] != 0 and all(board[row][check] == board[0][check] for row in range(num)):
            return board[0][check]

        check += 1

    if board[0][0] != 0 and all(board[i][i] == board[0][0] for i in range(num)):
        return board[0][0]

    if board[0][num - 1] != 0 and all(board[i][num - 1 - i] == board[0][num - 1] for i in range (num)):
        return board[0][num - 1]

    for row in board:
        for cell in row:
            if cell == 0:
                return -1

    return 0

def player_move(board, player, x, y):
    num = len(board)

    if x < 0:
        return False
    if x >= num:
        return False
    if y < 0:
        return False
    if y >= num:
        return False

    if board[x][y] != 0:
        return False

    board[x][y] = player
    return True        
    
def show_board(tac_grid):
    num = len(tac_grid)

    letters_def = ["A", "B", "C"]

    print("   " + "   ".join(letters_def[:num]))
    
    horizontal_cell = "  " + " ".join("---" for _ in range(num))
    print(horizontal_cell)

    row = 0

    while row < num:
        shown_row = []
        column = 0
        while column < num:
            if tac_grid[row][column] == 1:
                shown_row.append("X")
            elif tac_grid[row][column] == 2:
                shown_row.append("O")
            else:
                shown_row.append(" ")
                
            column += 1
        print(f"{row + 1} | " + " | ".join(shown_row) + " |")
        print(horizontal_cell)
        
        row += 1
    return True
    
def input_move():
    move = input("What move would you like to make? (Enter A1, B1, C1 etc.. Q or q to exit.)")

    if type(move) != str:
        return False
    if move == "Q" or move == "q":
        return "QUIT"
    if len(move) != 2:
        return False

    letters_def = ["A", "B", "C"]
    letter = move[0].upper()
    number = move[1]

    if letter not in letters_def:
        return False
    if not number.isdigit():
        return False

    x = int(number) - 1
    y = letters_def.index(letter)

    return x, y

def tic_tac_toe(tac_grid):
    try:
        num = len(tac_grid[0])
    except:
        print("Error")
        return False
        
    for row in tac_grid:
        try:
            if len(row) != num:
                print("Error")
                return False
        except:
            print("Error")
            return False
        for element in row:
            if type(element) != int:
                print("Error")
                return False
    try:
        if len(tac_grid) != num:
            print("Error")
            return False
    except:
        print("Error")
        return False

    players_turn = 1
    board_state = -1
    letters_def = ["A", "B", "C"]

    while board_state == -1:
        show_board(tac_grid)

        if players_turn == 1:
            print("X")
        else:
            print("O")

        move = input_move()
        if move == "QUIT":
            print("Exiting game.")
            return -1
        if move == False:
            print("Error")
            continue

        x, y = move

        if player_move(tac_grid, players_turn, x, y) == False:
            print("Error, bad move")
            continue

        board_state = check_state(tac_grid)

        if board_state == 0:
            show_board(tac_grid)
            print("This game is a draw.")
            return 0
            
        if board_state == 1:
            show_board(tac_grid)
            print("X wins!")
            return 1

        if board_state == 2:
            show_board(tac_grid)
            print("O wins!")
            return 2

        if players_turn == 1:
            players_turn = 2
        else:
            players_turn = 1

board = make_board(3)
tic_tac_toe(board)
````

````python
import copy

def make_board(num):
    try:
        if type(num) != int or num <= 0:
            print("Error")
            return False
    except:
        print("Error")
        return False

    board = []
    row = 0

    while row < num:
        board.append([])
        column = 0
        while column < num:
            board[row].append(0)
            column += 1
        row += 1
    return board

def check_state(board):
    num = len(board)

    check = 0
    while check < num:
        if board[check][0] != 0 and all(board[check][row] == board[check][0] for row in range(num)):
            return board[check][0]

        if board[0][check] != 0 and all(board[row][check] == board[0][check] for row in range(num)):
            return board[0][check]
        check += 1

    if board[0][0] != 0 and all(board[i][i] == board[0][0] for i in range(num)):
        return board[0][0]

    if board[0][num - 1] != 0 and all(board[i][num - 1 - i] == board[0][num - 1] for i in range (num)):
        return board[0][num - 1]

    for row in board:
        for cell in row:
            if cell == 0:
                return -1

    return 0

def player_move(board, player, x, y):
    num = len(board)

    if x < 0 or x >= num or y < 0 or y >= num:
        print("Not on this board, try another input!")
        return False

    if board[x][y] != 0:
        print("Spot is taken, try again.")
        return False

    board[x][y] = player
    return True        
    
def show_board(tac_grid):
    num = len(tac_grid)
    letters_def = ["A", "B", "C", "D", "E"]

    print("   " + "   ".join(letters_def[:num]))
    print(" " + " ---" * num)
    
    horizontal_cell = "  " + " ".join("---" for _ in range(num))
    print(horizontal_cell)

    for i in range(num):
        shown_row = []
        for j in range(num):
            if tac_grid[i][j] == 1:
                shown_row.append("X")
            elif tac_grid[i][j] == 2:
                shown_row.append("O")
            else:
                shown_row.append(" ")
            
        print(f"{i + 1} | " + " | ".join(shown_row) + " |")
        print(horizontal_cell)
        
    return True
    
def input_move(num):
    move = input("What move would you like to make? (Enter A1, B1, C1 etc.. Q or q to exit.)")

    if type(move) != str:
        return False
    if move == "Q" or move == "q":
        return "QUIT"
    if len(move) != 2:
        print("Must be on this board.")
        return False

    letters_def = ["A", "B", "C", "D", "E"]
    letter = move[0].upper()
    number = move[1]

    if letter not in letters_def[:num]:
        print("Must be in one of the columns of this board.")
        return False
        
    if not number.isdigit():
        print("There must be a row.")
        return False

    x = int(number) - 1
    
    if x < 0 or x >= num:
        print("Needs to be on this board.")
        return False
        
    y = letters_def.index(letter)
    return x, y

def score_board(board, player):
    state = check_state(board)
    if state == player:
        return 1
    if state == switch_player(player):
        return -1
    return 0

def switch_player(player):
    if player == 1:
        return 2
    else:
        return 1

def generate_moves(board, player, current_player, depth, keep_all_moves=False):
    if depth == 0 or check_state(board) != -1:
        return [], score_board(board, player)

    moves = []
    scores = []
    for i in range(len(board)):
        for j in range(len(board)):
            if board[i][j] == 0:
                new_board = copy.deepcopy(board)
                new_board[i][j] = current_player

                next_moves, score = generate_moves(new_board, player, switch_player(current_player), 
                                                   depth - 1, keep_all_moves = keep_all_moves)

                current_move = [(current_player, (i, j))]
                if keep_all_moves:
                    current_move.extend(next_moves)
                    
                moves.append(current_move)
                scores.append(score)
                
    return moves, scores

def tree_search(t, depth = 1):
    if isinstance(t, int):
        return t / depth
    if all(isinstance(x, int) for x in t):
        return sum(t) / depth
    return sum(tree_search(item, depth + 1) for item in t) / depth

def pick_move(board, player, depth = 3, func=max):
    moves, scores = generate_moves(board, player, player, depth)
    values = list(map(tree_search, scores))
    final_index = values.index(func(values))
    return moves[final_index][0][1]

def ai_vs_ai(size, depth):
    board = make_board(size)
    player = 1
    state = -1

    while state == -1:
        move = pick_move(board, player, depth)
        board[move[0]][move[1]] = player

        state = check_state(board)
        player = switch_player(player)
    return state

def ai_results(depth, games = 10):
    print("AI vs AI")
    print("--------------")

    for size in [3, 4, 5]:
        print(f"\nFor Board size: {size}x{size}")

        p1_wins = 0
        p2_wins = 0
        draws = 0

        
        for _ in range(games):
            result = ai_vs_ai(size, depth)
            if result == 1:
                p1_wins +=1
            elif result == 2:
                p2_wins += 1
            else:
                draws += 1

        print("Player 1 wins:", p1_wins)
        print("Player 2 wins:", p2_wins)
        print("Draws:", draws)

        print("Player 1 win rate:", p1_wins / games)
        print("Player 2 win rate:", p2_wins / games)    

        if p1_wins > p2_wins:
            print("Winner: Player 1")
        elif p2_wins > p1_wins:
            print("Winner: Player 2")
        else:
            print("Tie.")
        
        

def tic_tac_toe(tac_grid, ai_depth = 2):
    try:
        num = len(tac_grid[0])
    except:
        print("Error")
        return False

    try:
        if len(tac_grid) != num:
            print("Error")
            return False
    except:
        print("Error")
        return False

    for row in tac_grid:
        try:
            if len(row) != num:
                print("Error")
                return False
        except:
            print("Error")
            return False

        for element in row:
            if type(element) != int:
                print("Error")
                return False

    players_turn = 1
    board_state = -1
        
    while board_state == -1:
        show_board(tac_grid)

        if players_turn == 1:
            print("X")
            move = input_move(num)

            if move == "QUIT":
                print("Exiting game..")
                return -1
            if move is False:
                continue
            
            x, y = move
            if not player_move(tac_grid, 1, x, y):
                continue
        else:
            print("O")
            print("Computer is getting ready to make its move..")
            x, y = pick_move(tac_grid, 2, ai_depth)
            tac_grid[x][y] = 2

        board_state = check_state(tac_grid)

        if board_state == 1:
            show_board(tac_grid)
            print("X wins!")
            return 1
        elif board_state == 2:
            show_board(tac_grid)
            print("O wins!")
            return 2
        elif board_state == 0:
            show_board(tac_grid)
            print("The match has ended in a Draw.")
            return 0

        players_turn = switch_player(players_turn)

ai_results(depth = 3, games = 10)
````
