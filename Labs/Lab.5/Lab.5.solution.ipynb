```python
class Matrix:
    def __init__(self, n = None, m = None, data = None):
        if isinstance(n, int) and isinstance(m, int) and data is None:
            if n <= 0 or m <= 0:
                raise ValueError("The Matrix's dimensions have to be positive.")

            self.n = n
            self.m = m
            self.rows = n
            self.cols = m
            self.data = []

            for i in range(self.n):
                row = []

                for j in range(self.m):
                    row.append(0)

                self.data.append(row)
            return

        if data is None and isinstance(n, list) and m is None:
            data = n
            n = None

        if isinstance(data, list) and n is None and m is None:
            if not data or not all(isinstance(row,list) for row in data):
                raise ValueError("The Matrix should be initialized with a list of lists.")

            first = len(data[0])

            for row in data:
                if len(row) != first:
                    raise ValueError("All rows should have the same num of columns.")
            
            self.n = len(data)
            self.m = first
            self.rows = self.n
            self.cols = self.m
            self.data = []

            
            for row in data:
                next_row = []

                for value in row:
                    next_row.append(value)
                    
                self.data.append(next_row)
            return

        raise ValueError("Bad initialization")

    def shape(self):
        return (self.rows, self.cols)

    def to_list(self):
        return [row[:] for row in self.data]

    def transpose(self):
        t = []
        
        for j in range(self.cols):
            new_row = []
            
            for i in range(self.rows):
                new_row.append(self.data[i][j])
            t.append(new_row)
            
        return Matrix(data=t)

    def row(self, n):
        if not isinstance(n, int):
            raise TypeError("Row must be an integer.")
            
        return Matrix(data=[self.data[n][:]])

    def column(self, n):
        if not isinstance(n, int):
            raise TypeError("Column must be an integer.")
        col = []
        
        for i in range(self.rows):
            col.append([self.data[i][n]])
            
        return Matrix(data=col)

    def block(self, n_0, n_1, m_0, m_1):
        sub = []
        
        for i in range(m_0, m_1):
            sub_row = []
            
            for j in range(n_0, n_1):
                sub_row.append(self.data[i][j])
            sub.append(sub_row)
        return Matrix(data=sub)

    def __getitem__(self, key):
        if isinstance(key, tuple):
            if len(key) != 2:
                raise TypeError("You have to use M[i, j] with exactly two indices.")
            r, c = key

            if isinstance(r, int) and isinstance(c, int):
                return self.data[r][c]

            if isinstance(r, slice):
                row_idx = list(range(self.rows))[r]
            else:
                row_idx = [r]

            if isinstance(c, slice):
                col_idx = list(range(self.cols))[c]
            else:
                col_idx = [c]

            sub = []
            
            for i in row_idx:
                sub_row = []
                for j in col_idx:
                    sub_row.append(self.data[i][j])
                sub.append(sub_row)
                
            return Matrix(data=sub)

        return self.data[key]


    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            if len(key) != 2:
                raise TypeError("You have to use M[i, j] = value with exactly two indices.")
            i, j = key
            self.data[i][j] = value
            return

        if isinstance(key, slice) and key == slice(None):
            self.assign(value)
            return

        self.data[key] = value

    def assign(self, other):
        if isinstance(other, Matrix):
            if self.rows != other.rows or self.cols != other.cols:
                raise ValueError("The matrices must have same dimensions.")
            for i in range(self.rows):
                for j in range(self.cols):
                    self.data[i][j] = other.data[i][j]
            return

        if isinstance(other, list):
            if len(other) != self.rows or any(len(row) != self.cols for row in other):
                raise ValueError("List dimensions must match the matrix's dimensions.")
            for i in range(self.rows):
                for j in range(self.cols):
                    self.data[i][j] = other[i][j]
            return

        raise TypeError("Bad assignment type")

    def __repr__(self):
        s = ""
        for row in self.data:
            s += str(row) + "\n"
        return s

    def constant(n, m, c):
        if not isinstance(n, int) or not isinstance(m, int):
            raise TypeError("n and m must be integers.")

        if n <= 0 or m <= 0:
            raise ValueError("n and m must be positive")
        data = []

        for i in range(n):
            row = []

            for j in range(m):
                row.append(float(c))

            data.append(row)
        return Matrix(data=data)

    def zeros(n, m):
        return Matrix.constant(n, m, 0.0)

    def ones(n, m):
        return Matrix.constant(n, m, 1.0)

    def eye(n):
        if not isinstance(n, int):
            raise TypeError("n must be an integer.")
            
        if n <= 0:
            raise ValueError("n must be positive.")
            
        data = []

        for i in range(n):
            row = []

            for j in range(n):
                row.append(1.0 if i == j else 0.0)

            data.append(row)
        return Matrix(data=data)

    def scalarmul(self, c):
        filled_matrix = []
        
        for i in range(self.rows):
            row = []
            
            for j in range(self.cols):
                row.append(self.data[i][j] * c)
                
            filled_matrix.append(row)
        return Matrix(data=filled_matrix)

    def add(self, N):
        if not isinstance(N, Matrix):
            raise TypeError("N must be a Matrix.")
            
        if self.rows != N.rows or self.cols != N.cols:
            raise ValueError("Matrices should have the same dimensions for addition.")

        filled_matrix = []
        
        for i in range(self.rows):
            row = []
            
            for j in range(self.cols):
                row.append(self.data[i][j] + N.data[i][j])
                
            filled_matrix.append(row)
        return Matrix(data=filled_matrix)

    def sub(self, N):
        if not isinstance(N, Matrix):
            raise TypeError("N must be a Matrix.")
            
        if self.rows != N.rows or self.cols != N.cols:
            raise ValueError("Matrices should have the same dimensions for subtraction")

        filled_matrix = []
        
        for i in range(self.rows):
            row = []
            
            for j in range(self.cols):
                row.append(self.data[i][j] - N.data[i][j])
                
            filled_matrix.append(row)
        return Matrix(data=filled_matrix)

    def element_mult(self, N):
        if not isinstance(N, Matrix):
            raise TypeError("N must be a Matrix.")
            
        if self.rows != N.rows or self.cols != N.cols:
            raise ValueError("Matrices should have the same dimensions for element-wise multiplication.")

        filled_matrix = []
        
        for i in range(self.rows):
            row = []
            
            for j in range(self.cols):
                row.append(self.data[i][j] * N.data[i][j])
                
            filled_matrix.append(row)
        return Matrix(data=filled_matrix)

    def mat_mult(self, N):
        if not isinstance(N, Matrix):
            raise TypeError("N must be a Matrix.")
            
        if self.cols != N.rows:
            raise ValueError("Inner dimensions should match for matrix multipicatn.")

        filled_matrix = []
        
        for i in range(self.rows):
            row = []
            
            for j in range(N.cols):
                total = 0
                
                for k in range(self.cols):
                    total += self.data[i][k] * N.data[k][j]
                row.append(total)
                
            filled_matrix.append(row)
        return Matrix(data=filled_matrix)

    def equals(self, N):
        if not isinstance(N, Matrix):
            return False
            
        if self.rows != N.rows or self.cols != N.cols:
            return False
            
        for i in range(self.rows):
            for j in range(self.cols):
                if self.data[i][j] != N.data[i][j]:
                    return False
                    
        return True

    def __add__(self, other):
        return self.add(other)

    def __sub__(self, other):
        return self.sub(other)

    def __mul__(self, other):
        if isinstance(other, (int, float)):
            return self.scalarmul(other)
            
        if isinstance(other, Matrix):
            return self.mat_mult(other)
            
        raise TypeError("Multiplication not supported")

    def __rmul__(self, other):
        if isinstance(other, (int, float)):
            return self.scalarmul(other)
            
        raise TypeError("Multiplication not supported")

    def __eq__(self, other):
        return self.equals(other)


A = Matrix([[2, -1],
            [3,  4]])
B = Matrix([[0,  1],
            [1,  2]])
C = Matrix([[-2, 5],
            [ 1, 0]])

print("Testing Matrix A", A)
print("Testing Matrix B", B)
print("Testing Matrix C", C)
print("---------------------------")

print("(AB)C = A(BC)")

print("Testing (A * B) * C", (A * B) * C)
print("Testing A * (B * C)", A * (B * C))
print("Testing ((A * B) * C) == (A * (B * C))", ((A * B) * C) == (A * (B * C)))

print("A(B + C) = AB + AC")
print("Testing A * (B + C)", A * (B + C))
print("Testing (A * B) + (A * C)", (A * B) + (A * C))
print("Testing (A * (B + C)) == ((A * B) + (A * C))", (A * (B + C)) == ((A * B) + (A * C)))

print("AB != BA")
print("Testing A * B", A * B)
print("Testing B * A", B * A)
print("Testing (A * B) == (B * A)", (A * B) == (B * A))

print("AI = A")
I = Matrix.eye(2)
print("Testing the Identity Matrix, I", I)
print("Testing A * I", A * I)
print("Testing (A * I) == A", (A * I) == A)
```
